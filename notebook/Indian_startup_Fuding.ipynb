# Dataset Summary

The **Indian Startup Funding** dataset contains information about startup funding events across India. It includes details such as the funding date, startup name, industry, city, investors, investment type, funding amount, and remarks. The dataset is useful for analyzing investment patterns, identifying funding trends, understanding investor behaviour, and exploring the growth of the Indian startup ecosystem.

The dataset consists of **3,044 rows** and **10 columns**, where each row represents a single funding event.

---

```python
df.drop_duplicates(inplace=True)
```

Rename Columns

```python
df.columns=[
    "Sr_No",
    "Date",
    "Startup_Name",
    "Industry_Vertical",
    "SubVertical",
    "City",
    "Investors_Name",
    "Investment_Type",
    "Amount_USD",
    "Remarks"
]
```

Convert Date

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)
```

```python
df.info()
```

Extract

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month_name()
df["Quarter"] = df["Date"].dt.quarter
```

```python
df.head()
```

Clean Amount Column

```python
df["Amount_USD"].sample(20)
```

Remove commas

```python
df["Amount_USD"] = df["Amount_USD"].astype(str)
df["Amount_USD"] = df["Amount_USD"].str.replace(",","")
```

Replace Unknown Values

```python
import numpy as np

df["Amount_USD"] = df["Amount_USD"].replace(
    ["undisclosed", "Undisclosed", "unknown", "Unknown"],
    np.nan
)
```

```python
df.sample(10)
```

Convert to Numeric

```python

df["Amount_USD"] = pd.to_numeric(df["Amount_USD"], errors="coerce")
```

Standardize Text

```python
df["City"]=df["City"].str.strip().str.title()
df["Startup_Name"]=df["Startup_Name"].str.strip()

df["Industry_Vertical"]=df["Industry_Vertical"].str.strip()

df["Investment_Type"]=df["Investment_Type"].str.strip()
df["Industry_Vertical"] = df["Industry_Vertical"].str.strip()

df["Industry_Vertical"] = df["Industry_Vertical"].replace({
    "eCommerce": "E-Commerce",
    "ECommerce": "E-Commerce",
    "Ecommerce": "E-Commerce",
    "e-commerce": "E-Commerce",
    "E-Commerce": "E-Commerce"
})
# Clean industry names
df["Industry_Vertical"] = df["Industry_Vertical"].str.strip()

# Standardize common variations
df["Industry_Vertical"] = df["Industry_Vertical"].replace({
    "FinTech": "Fintech",
    "fintech": "Fintech",
    "Ecommerce": "E-Commerce",
    "eCommerce": "E-Commerce",
    "ECommerce": "E-Commerce",
    "e-commerce": "E-Commerce",
    "E commerce": "E-Commerce"
})
```

Handaling Missing Values


```python
df.isnull().sum()
```

```python
df["City"] = df["City"].fillna("Unknown")

df["Industry_Vertical"] = df["Industry_Vertical"].fillna("Unknown")

df["Amount_USD"] = df["Amount_USD"].fillna(df["Amount_USD"].median())
df["Investors_Name"] = df["Investors_Name"].replace({
    "Undisclosed investors": "Undisclosed Investors"
})
```

```python
# Clean Investment_Type column

df["Investment_Type"] = (
    df["Investment_Type"]
    .astype(str)
    .str.strip()
    .str.replace(r"\\\\n", " ", regex=True)   # Seed\\nFunding -> Seed Funding
    .str.replace(r"\s+", " ", regex=True)
)

# Standardize categories
df["Investment_Type"] = df["Investment_Type"].replace({

    # Private Equity
    "Private Equity Round": "Private Equity",
    "Private": "Private Equity",
    "Equity": "Private Equity",
    "Equity Based Funding": "Private Equity",

    # Seed Funding
    "Seed": "Seed Funding",
    "Seed Round": "Seed Funding",
    "Seed Funding Round": "Seed Funding",
    "Seed funding": "Seed Funding",
    "Seed Funding": "Seed Funding",

    # Seed / Angel Funding
    "Seed/ Angel Funding": "Seed / Angel Funding",
    "Seed/Angel Funding": "Seed / Angel Funding",
    "Angel / Seed Funding": "Seed / Angel Funding",
    "Seed / Angle Funding": "Seed / Angel Funding",

    # Pre-Series
    "pre-Series A": "Pre-Series A",
    "Pre Series A": "Pre-Series A",

    # Venture
    "Venture Round": "Venture Funding",

    # Debt
    "Debt-Funding": "Debt Funding",
    "Debt and Preference capital": "Debt Funding",

    # Crowdfunding
    "Crowd Funding": "Crowdfunding",
    "Crowd funding": "Crowdfunding"
})

# Group rare categories (optional)
counts = df["Investment_Type"].value_counts()

df["Investment_Type"] = df["Investment_Type"].apply(
    lambda x: x if counts[x] >= 5 else "Others"
)

print(df["Investment_Type"].value_counts())
```

# 📊 Business Insights

After cleaning and analyzing the Indian Startup Funding dataset, several key business insights were identified:

### 1. Funding is concentrated in a few major cities
- Bengaluru, Mumbai, and New Delhi account for the largest share of startup funding.
- These cities have established startup ecosystems, better access to investors, and stronger infrastructure for entrepreneurship.

### 2. Consumer-focused industries attract significant investment
- Industries such as E-commerce, Consumer Internet, FinTech, and Technology received the highest funding.
- Investors tend to prefer sectors with high growth potential and large customer bases.

### 3. Private Equity and Seed Funding dominate the market
- Private Equity and Seed Funding represent the majority of funding rounds.
- This indicates strong investor interest in both early-stage startups and rapidly growing companies.

### 4. Startup funding has changed over the years
- The number of funding rounds and total investment vary significantly from year to year.
- Certain years experienced rapid growth due to increased investor confidence and expansion of the Indian startup ecosystem.

### 5. A few investors participate in a large number of funding rounds
- Some investors consistently invest across multiple startups, making them key contributors to startup growth.
- These investors play an important role in supporting innovation and entrepreneurship.

### 6. Funding distribution is highly skewed
- A small number of startups receive exceptionally large investments, while the majority receive relatively smaller funding amounts.
- This creates noticeable outliers in the funding data.

### 7. Data preprocessing is essential
- The raw dataset contained missing values, inconsistent naming conventions, duplicate records, and non-numeric funding amounts.
- Proper data cleaning significantly improved the quality and reliability of the analysis.
